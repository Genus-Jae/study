```python
# import 구문은 제일 위에 모아두는 거 기억하시죠?
import urllib.parse
import urllib.request
import json
import mysql.connector
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
!pip install mysql-connector-python
```

```python
client_id = os.getenv("naver_client_id")
client_secret = os.getenv("naver_secret")
```

```python
#검색어 설정
searchText = urllib.parse.quote('한국축구')
```

```python
#url 및 헤더 설정
url= "https://openapi.naver.com/v1/search/book.json?query=" + searchText + "&dispaly=100"


#네이버 API 요청 준비
request = urllib.request.Request(url)
#인증 헤더 추가
request.add_header("X-Naver-Client-Id",client_id)
request.add_header("X-Naver-Client-Secret",client_secret)
```

```python
#api 요청 및 결과값 json 반환
#[힌트] json <-> 파이썬 ditionary 변환
# import json
# json.loads(json)
response = urllib.request.urlopen(request) #api 호출
response_body = response.read()            #응답데이터
response_body = json.loads(response_body)  #json -> dictionary
book_list = response_body['items']         #책 데이터 목록
```

```python
book_list
```

```python
# DB 연결 객체 생성
with mysql.connector.connect(
    host = "localhost",
    user = "capybara",
    password = "1234",
    database = "bookdb"
) as connection:
    
    # SQL 수행을 위한 커서 생성
    with connection.cursor() as cursor:

        sql = "insert into naver_book" \
            " (book_title, book_image, author, publisher, isbn, book_description, pub_date)" \
            " values (%s,%s,%s,%s,%s,%s,%s)"

        for book_info in book_list:
            values = (

                book_info['title'],
                book_info['image'],
                book_info['author'],
                book_info['publisher'],
                book_info['isbn'],
                book_info['description'],
                book_info['pubdate'],     
            )

            cursor.execute(sql, values)   #INSERT문 실행
            
    # 데이터베이스 트랜잭션 처리 (commit -> 반영)        
    connection.commit() # 실제 DB상에 저장    

sql
```
