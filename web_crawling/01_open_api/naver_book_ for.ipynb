```python
client_id = os.getenv("naver_client_id")
client_secret = os.getenv("naver_secret")
```

```python
#검색어 설정
searchText = urllib.parse.quote('야구')
```

```python
#url 및 헤더 설정
url= "https://openapi.naver.com/v1/search/book.json?query=" + searchText + "&dispaly=100"


#네이버 API 요청 준비
request = urllib.request.Request(url)
#인증 헤더 추가
request.add_header("X-Naver-Client-Id",client_id)
request.add_header("X-Naver-Client-Secret",client_secret)
```

```python
#api 요청 및 결과값 json 반환
#[힌트] json <-> 파이썬 ditionary 변환
# import json
# json.loads(json)
response = urllib.request.urlopen(request) #api 호출
response_body = response.read()            #응답데이터
response_body = json.loads(response_body)  #json -> dictionary
response_body      
```

```python
total_count = response_body['total'] #전체 검색 결과 개수
start_num = 1
loop_count = total_count // 100 + 1 #100개씩 끊어서 몇 번 반복할지 계산
book_list = [] # 모든 책 데이터를 누적할 리스트
```

```python
for i in range(loop_count):
    #url 및 헤더 설정
    url= "https://openapi.naver.com/v1/search/book.json?query=" + searchText + "&dispaly=100&start=" + str(start_num)


    #네이버 API 요청 준비
    request = urllib.request.Request(url)
    #인증 헤더 추가
    request.add_header("X-Naver-Client-Id",client_id)
    request.add_header("X-Naver-Client-Secret",client_secret)

    response = urllib.request.urlopen(request) #api 호출
    response_body = response.read()            #응답데이터
    response_body = json.loads(response_body)  #json -> dictionary

    book_list += response_body ['items']       #rlwhs fltmxmdp dldjqnxdla
    start_num += 100

    #최대 1000건만 요청
    if start_num > 1000:
        break

len(book_list)    
```
