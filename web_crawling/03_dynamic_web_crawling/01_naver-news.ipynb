```python
from selenium import webdriver
from selenium.webdriver.common.by import By  #요소 선택용
from selenium.webdriver.common.keys import Keys  #키보드 이벤트 제어
import time
```

```python
#Chrome 브라우저 실행
service = webdriver.chrome.service.Service('chromedriver.exe')  #Service 객체 생성
driver = webdriver.Chrome(service=service)                      #Chrome 브라우저 실행
```

```python
#특정 url 접근

driver.get("https://naver.com")
time.sleep(1)
```

```python
search_box = driver.find_element(By.ID, 'query')
search_box.send_keys('이번주 날씨')
search_box.send_keys(Keys.RETURN)
time.sleep(1)
```

```python
news_tab = driver.find_element(By.CSS_SELECTOR,"a.tab[href*='where=news']") #뉴스탭 찾기
news_tab.click()
time.sleep(1)
```

```python
for _ in range(5):
    body = driver.find_element(By.TAG_NAME, "body")
    body.send_keys(Keys.END)
    time.sleep(1)
```

```python
news_contents = driver.find_elements(By.CSS_SELECTOR, 'span.sds-comps-text-type-headline1')

for news_content in news_contents:
    href_tag = news_content.find_element(By.XPATH,"..")



    title = news_content.text #XPATH를 이용한 태그 탐색 (a 태그)

    title = news_content.text                  #span 태그 : 제목
    href = href_tag.get_attribute('href')      # a 태그 : 링크 경로

    print(title, "ㅣ",href)

#드라이버 종료 (브라우저 종료)
driver.quit()
```
