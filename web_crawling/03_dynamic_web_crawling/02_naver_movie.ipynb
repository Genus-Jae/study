```python
# 셀레니움 관련 라이브러리

from selenium import webdriver
from selenium.webdriver.common.by import By  #요소 선택용
from selenium.webdriver.common.keys import Keys  #키보드 이벤트 제어
from selenium.webdriver.chrome.service import Service


# 시간관련 라이브러리
import time
from datetime import datetime          # 파일명 날짜시간

# 다운로드 관련 라이브러리
from urllib.request import urlretrieve # 이미지 다운로드 
```

```python
import mysql.connector

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="1234",
    database="python_test"
)

if connection.is_connected():
    print("mysql에 성공적으로 연결되었습니다.")

connection.close()
```

```python
#Chrome 브라우저 실행
service = webdriver.chrome.service.Service('chromedriver.exe')  #Service 객체 생성
driver = webdriver.Chrome(service=service)                      #Chrome 브라우저 실행
```

```python
#특정 url 접근

driver.get("https://naver.com")
time.sleep(1)
```

```python
#검색어 입력 및 검색

search_box = driver.find_element(By.ID, 'query')
search_box.send_keys('현재 상영영화')
search_box.send_keys(Keys.RETURN)
time.sleep(1)
```

```python
for i in range(9):
    movie_items = driver.find_elements(By.CSS_SELECTOR, 'div.data_area')

    for movie_item in movie_items:
        #제목이 들어있는 a 태그 글자만 추출
        title = movie_item.find_element(By.CSS_SELECTOR, 'div.area_text_box > a').text
        
        #이미지 + 링크 포함된 요소
        img_box = movie_item.find_element(By.CSS_SELECTOR, 'a.img_box')
        
        #해당 영화 상세페이지 링크
        link = img_box.get_attribute('href')
        
        #이미지 url
        img_src = img_box.find_element(By.TAG_NAME,"img").get_attribute('src')

        print(f"제목 : {title}, 상세페이지 : {link}, 이미지 url : {img_src}")

        file_name = f"../movie_images/{datetime.now().strftime('%y%m%d_%H%M%S_')}_{title}.jpg"
        urlretrieve(img_src, file_name) # 파일 저장 (파일, 저장경로와 파일명)

    #다음 페이지로 이동
    if i != 8:
        next_btn = driver.find_element(By.CSS_SELECTOR , 'a._next')
        next_btn.click()
        time.sleep(1)


driver.quit
```
