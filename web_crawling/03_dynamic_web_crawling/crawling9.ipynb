```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome()
driver.get("https://www.opinet.co.kr/user/cufaq/cufaqSelect.do")
time.sleep(2)

# [Step 1] 전체 링크를 먼저 확보하기 (목록에서 링크만 추출)
all_links = []
for page_num in range(1, 5):
    driver.execute_script(f"fn_egov_link_page({page_num})")
    time.sleep(1)
    # 상세 페이지로 가는 a 태그의 href 속성을 모두 가져옴
    elements = driver.find_elements(By.CSS_SELECTOR, "td.t_left.input a")
    for el in elements:
        all_links.append(el.get_attribute("href"))

# [Step 2] 확보된 링크를 하나씩 접속해서 데이터 뽑기
data_list = []
for link in all_links:
    driver.get(link)  # 뒤로가기 대신 링크로 바로 이동! (이게 궁딸깍!)
    time.sleep(0.5)
    
    title = driver.find_element(By.CSS_SELECTOR, "div.board_view_tit").text
    content = driver.find_element(By.CLASS_NAME, "view_contents").text
    
    data_list.append({"제목": title, "내용": content})
    print(f"수집 완료: {title}")

# 저장
df = pd.DataFrame(data_list)
df.to_excel("../data/faq_data.xlsx", index=False)
driver.quit()
```
