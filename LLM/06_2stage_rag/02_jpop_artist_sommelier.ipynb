## 취향에 맞는 J-POP 가수 추천

```python
system_prompt = '''
당신은 일본 대중음악을 폭넓게 듣고 분석하는 J-POP 가수 추천 소믈리에입니다.
사용자가 좋아하는 가수, 노래, 장르 또는 분위기를 분석하여 어울리는 J-POP 가수 3명을 추천하세요.

각 가수마다 다음 내용을 한국어로 간단히 작성하세요.
- 가수 이름
- 입문용 대표곡 1곡
- 추천 이유

같은 가수를 중복 추천하지 말고, 확인되지 않은 곡이나 사실은 만들어내지 마세요.
'''
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ('system', system_prompt),
    ('human', '다음 음악 취향에 어울리는 J-POP 가수를 추천해 주세요.\n취향: {preference}')
])

model = init_chat_model('gpt-5.6-luna')
output_parser = StrOutputParser()

chain = prompt | model | output_parser

response = chain.invoke('신나는')
print(response)
```

```python
def recommend_jpop_artist(preference: str) -> str:
    return chain.invoke({
        'preference': preference
    })
```

```python
response = recommend_jpop_artist(
    '요네즈 켄시와 킹누를 좋아하고, 감성적이면서 밴드 사운드가 강한 노래를 좋아해요.'
)

print(response)
```

```python
response = recommend_jpop_artist(
    'Novelbright 대표곡 3개 알려줘.'
)

print(response)
```
