```python
import matplotlib.pyplot as plt

#데이터 준비
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [22,24,19,23,25,20,18]

plt.plot(days,           #x축 days
         temperatures, 
         marker = 'o',   #데이터 마커
         linestyle = '-', #선 스타일
         color = 'y')

plt.title('Temperature Variation Over a Week')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.legend()
```

```python
import matplotlib.pyplot as plt

#데이터 준비
months = ['Jan','Feb', 'Mar', 'Apr', 'May']
sales = [150,200,180,220,170]

# 막대 그래프 그리기
plt.plot(months, sales, color = 'r', marker = '^')
plt.bar(months, sales, color='skyblue')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales Amount')

for i, month in enumerate(months):
    plt.text(month , sales[i] + 5, sales[i])
plt.show()    
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 데이터 준비
heights = np.random.normal(170, 5, 200)  # 평균 170cm, 표준편차 5cm인 데이터 200개 생성

plt.hist(heights, bins = 10, color = 'green', edgecolor = 'red')
plt.title('Height Distribution of Students')
plt.xlabel('Height (cm)')
plt.ylabel('Number of Students')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 데이터 준비
np.random.seed(0)
study_hours = np.random.normal(5, 1.5, 100)  # 평균 5시간, 표준편차 1.5시간
scores = study_hours * 10 + np.random.normal(0, 5, 100)  # 공부 시간에 따른 점수

plt.scatter(study_hours, scores, color = 'purple', alpha = 0.6)
plt.title('Relationship between Study Hours and Exam Scores')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.show()
```

```python
import matplotlib.pyplot as plt

# X축과 Y축 데이터 설정
x = [1, 2, 3, 4]
y1 = [10, 20, 25, 30]
y2 = [30, 25, 20, 10]
y3 = [20, 30, 10, 40]

# 다양한 선 스타일로 그래프 그리기
plt.plot(x, y1, linestyle='-', label='Solid Line')
plt.plot(x, y2, linestyle='--', label='Dashed Line')
plt.plot(x, y3, linestyle=':', label='Dotted Line')

# 레이블 및 타이틀 추가
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Line Styles Example')

# 범례 추가
plt.legend()

# 그래프 보여주기
plt.show()
```

```python
# RGB 값으로 색상 지정
plt.plot(x, y1, color=(0.1, 0.2, 0.5), label='RGB Color')

# HEX 코드로 색상 지정
plt.plot(x, y2, color='#FF5733', label='Hex Color')
```

```python
import matplotlib.pyplot as plt

# X축 데이터 설정
x = [1, 2, 3, 4, 5]

# Y축 데이터 설정
y1 = [1, 4, 9, 16, 25]
y2 = [25, 16, 9, 4, 1]

# 첫 번째 선: 빨간 실선, 투명도 설정
plt.plot(x, y1, label='y = x^2', color='r', alpha=0.3)

# 두 번째 선: 파란 점선, 투명도 설정
plt.plot(x, y2, label='y = -x^2', color='b', alpha=0.8)

# 레이블 및 타이틀 추가
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Multiple Lines with Transparency')

# 범례 추가
plt.legend()

# 그래프 보여주기
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 데이터 준비
x = np.linspace(0, 10, 100)  #간격 구분

# 서브플롯 생성
plt.figure(figsize=(10, 5))

# 첫 번째 그래프
plt.subplot(1, 2, 1)   #1행 2열 중 첫 번째
plt.plot(x, np.sin(x))
plt.title('Sine Wave')

# 두 번째 그래프
plt.subplot(1, 2, 2)   #1행 2열 중 두 번째
plt.plot(x, np.cos(x))
plt.title('Cosine Wave')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 2D 배열 데이터 생성
data = np.random.rand(10, 10)

# 두 개의 서브플롯에서 다른 컬러맵 적용
plt.subplot(1, 2, 1)
plt.imshow(data, cmap='plasma')
plt.title('Plasma Colormap')
plt.colorbar()

plt.subplot(1, 2, 2)
plt.imshow(data, cmap='inferno')
plt.title('Inferno Colormap')
plt.colorbar()

# 서브플롯 간격 조정
plt.tight_layout()

# 그래프 보여주기
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

# 사용자 정의 컬러맵 설정
# LinearSegmentedColormap.from_list()를 사용하여 색상 리스트를 기반으로 컬러맵 정의
colors = ['#0000FF', '#00FF00', '#FF0000']  # 파란색에서 녹색, 빨간색으로 변환
custom_cmap = LinearSegmentedColormap.from_list('custom', colors)

# 2D 배열 데이터 생성
data = np.random.rand(10, 10)

# 사용자 정의 컬러맵 적용
plt.imshow(data, cmap=custom_cmap)
plt.colorbar()

# 타이틀 추가
plt.title('Custom Colormap')

# 그래프 보여주기
plt.show()
```
