## **[종합 실습 1] 상품 판매 데이터 분석**

다음은 한 전자상거래 사이트의 **상품 판매 데이터**입니다. 고객들이 특정 상품을 구매한 기록이 포함되어 있습니다. 데이터를 활용하여 아래의 문제를 해결하세요.

```python
import pandas as pd

# 예시 상품 판매 데이터
sales_data = {
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Mouse', 'Laptop', 'Monitor', 'Mouse', 'Keyboard', 'Laptop', 'Mouse', 'Monitor', 'Keyboard'],
    'Quantity': [1, 2, 1, 1, 2, 1, 1, 3, 1, 1],
    'Price': [1200, 25, 1200, 300, 25, 100, 1200, 25, 300, 100],
    'CustomerID': [1001, 1002, 1001, 1003, 1004, 1002, 1001, 1003, 1005, 1002],
    'OrderDate': ['2026-01-05', '2026-01-06', '2026-01-07', '2026-01-10', '2024-01-12', '2026-01-15', '2026-01-18', '2026-01-20', '2026-01-22', '2026-01-25']
}
df_sales = pd.DataFrame(sales_data)
```

1. 총 구매금액 컬럼 Total(Quantity * Price)을 만들고, 각 고객(CustomerID)별로 총 구매 금액을 계산하세요.
2. `Laptop`을 구매한 고객들의 평균 구매 금액을 계산하세요.
3. 가장 많이 판매된 상품(상품 이름과 판매된 총 수량)을 구하세요.
4. `OrderDate`가 2026년 1월 10일 이후인 주문만 필터링하여 출력하세요.
5. `Product`별로 판매량과 총 판매 금액을 계산한 후, 이를 `총 판매금액` 기준으로 내림차순으로 정렬하여 출력하세요.

````python
import pandas as pd

# 예시 상품 판매 데이터
sales_data = {
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Mouse', 'Laptop', 'Monitor', 'Mouse', 'Keyboard', 'Laptop', 'Mouse', 'Monitor', 'Keyboard'],
    'Quantity': [1, 2, 1, 1, 2, 1, 1, 3, 1, 1],
    'Price': [1200, 25, 1200, 300, 25, 100, 1200, 25, 300, 100],
    'CustomerID': [1001, 1002, 1001, 1003, 1004, 1002, 1001, 1003, 1005, 1002],
    'OrderDate': ['2026-01-05', '2026-01-06', '2026-01-07', '2026-01-10', '2024-01-12', '2026-01-15', '2026-01-18', '2026-01-20', '2026-01-22', '2026-01-25']
}

df_sales = pd.DataFrame(sales_data)

df_sales
````

````python
df_sales['Quantity'] * df_sales['Price']
````

````python
# 총 구매금액 컬럼 Total(Quantity * Price)을 만들고, 각 고객(CustomerID)별로 총 구매 금액을 계산하세요.
df_sales['Total'] = df_sales['Quantity'] * df_sales['Price']

customer_total = df_sales.groupby('CustomerID')['Total'].sum()
customer_total
````

````python
df_sales[df_sales['Product'] == 'Laptop']
````

````python
# `Laptop`을 구매한 고객들의 평균 구매 금액을 계산하세요.
laptop_sales = df_sales[df_sales['Product'] == 'Laptop']
laptop_average = laptop_sales['Total'].mean()
print(laptop_average)
````

````python
df_sales.groupby('Product')['Quantity'].sum()
````

````python
# 가장 많이 판매된 상품(상품 이름과 판매된 총 수량)을 구하세요.
product_sales = df_sales.groupby('Product')['Quantity'].sum()

# product_sales.max()  # 7
# product_sales.idxmax()  # Mouse
best_product_sales = product_sales.max()
best_product = product_sales.idxmax()
print(f"가장 많이 판매된 제품 : {best_product}, 판매된 수량 : {best_product_sales}")
````

````python
# `OrderDate`가 2026년 1월 10일 이후인 주문만 필터링하여 출력하세요.
df_sales[df_sales['OrderDate'] > '2026-01-10']
````

````python
df_sales['OrderDate'] = pd.to_datetime(df_sales['OrderDate'])
filtered_sales = df_sales[df_sales['OrderDate'] > '2026-01-10']
filtered_sales
````

````python
# `Product`별로 판매량과 총 판매 금액을 계산한 후,
# 이를 `총 판매금액` 기준으로 내림차순으로 정렬하여 출력하세요.
product_total = df_sales.groupby('Product').agg({'Quantity' : 'sum', 'Total' : 'sum'})
sorted_product_total = product_total.sort_values('Total', ascending=False)
sorted_product_total
````
