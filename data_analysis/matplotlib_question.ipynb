## [실습 1] 다양한 그래프 활용 1

- 주어진 데이터에 대해 선 그래프, 막대 그래프, 히스토그램, 산점도를 한꺼번에 그려보세요.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)

# 데이터 준비
x = np.arange(1, 11)
y = np.random.randint(10, 100, 10)

data = np.random.normal(50, 15, 100)
```

```python
plt.plot(data)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
# 막대 그래프 그리기
x = np.arange(1, 11)
y = np.random.randint(10, 100, 10)
plt.bar(x, y, color = 'red')
plt.title('x and y')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# 데이터 준비
x = np.arange(1, 11)
y = np.random.randint(10, 100, 10)

# 산점도 그리기
plt.scatter(x, y, color='purple', alpha=0.6)
plt.title('Relationship between Study Hours and Exam Scores')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np  # numpy 직접 가져오기

# 예제 데이터 생성
months = ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'November', 'December']
avg_temperatures = [-1, 0, 5, 10, 15, 20, 25, 23, 19, 12, 5, 0]
temperatures = pd.DataFrame({'Month': months, 'Avg': avg_temperatures })

sns.lineplot(x='Month', y='Avg', data=temperatures)
plt.title('Simple month Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터셋 로드
months = ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'November', 'December']
rainfall = [50, 40, 60, 80, 100, 120, 150, 140, 110, 90, 70, 60]
rainfalls = pd.DataFrame({'Month': months, 'Rain': rainfall })
# 막대 그래프 그리기
sns.barplot(x='months', y='rainfall', data=rainfalls)
plt.title('Total Tips by Day')
plt.xlabel('Day of Week')
plt.ylabel('Total Tip Amount')
plt.show()
```
