```python
address = '대한민국 서울시 서초구'
# 서초구 (9번 인덱스부터 끝까지)
#서초구 (뒤에서 3번재 인덱스부터 끝까지)
#한서서 (1번 인덱스부터 12번 인덱스 전까지 추출하는데 4개씩 건너뛰면서 추출)
#구초서 서울서 국민한대 (슬라이싱을 이용해 [::-1]을 주면 문자열 뒤집기)
print(address[9:])
print(address[-3:])
print(address[1::4])
print(address[::-1])
```
