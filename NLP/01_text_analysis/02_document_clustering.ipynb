````python
import numpy as np
import pandas as pd
import os, glob     # 운영체제 경로 처리 / 파일 패턴 
path = "../data/OpinionsDataset1.0/topics"
all_files = glob.glob(os.path.join(path, '*.data'))

filename_list = []
opinions_list = []

for file_ in all_files:
    df = pd.read_table(
        file_,               # 경로
        header = None,       # 첫 줄은 컬럼명으로 사용안함
        index_col = None,    # 인덱스 컬럼 지정 안함(기본생성)
        encoding = "latin1"  # 특수문자 깨짐 방지
    )

    display(df)

    filename = os.path.basename(file_)     ## 파일명 추출
    filename = filename.split('.')[0]      # 확장자 제거
    filename_list.append(filename)         # 파일명 리스트 추가

    opinions = df.to_string(index = False, header = False) #인덱스/헤더 제거
    opinions_list.append(opinions)
````

````python
document_df = pd.DataFrame({
    'filename': filename_list,
    'opinions': opinions_list
})

document_df
````

````python
import nltk

nltk.download('wordnet') # 대형 의미 사전
````

````python
from nltk.stem import WordNetLemmatizer  # 표제어(lemma) 추출 도구

lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize('running', pos = 'v')) # 동사로 간주해 표제어(사전 대표 기본형) 변환
print(lemmatizer.lemmatize('ran', pos = 'v'))
````

````python
# 파이썬에서 사용하는 문장부호 목록
import string
string.punctuation
````

````python
# 텍스트를 정리하고 표제어(동사 기준)으로 반환하는 함수
def lemmatize(text):
    text = text.lower() # 소문자 변환

    punc_rem_dict = dict((ord(ch), None) for ch in string.punctuation) # 문장부호 제거용 치환 딕셔너리
    print(punc_rem_dict)
    text = text.translate(punc_rem_dict)  # text에서 문장부호에 해당하는 것들은 None 매핑 -> 삭제처리

    tokens = nltk.word_tokenize(text)     # 단어 단위로 토큰화

    lemmatizer = WordNetLemmatizer()      # 표제어 추출 객체 생성
    return [lemmatizer.lemmatize(token, pos = 'v') for token in tokens]

lemmatize('The Matrix is everywhere its all around us, here even in this room. You can see it out your')
````

````python
# Lemmatize 적용한 후 TF-IDF 벡터화
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(
    tokenizer = lemmatize,          # 사용자 정의 토크나이저 적용
    stop_words = 'english',         # 영어 불용어 제거
    ngram_range = (1,2),            # 유니그램, 바이그램
    max_df = 0.85,                  # 전체 문서 85% 초과 등장시 흔한 단어이므로 제외
    min_df = 0.05                   # 전체 문서 5% 미만 등장시 희귀한 단어이므로 제외
)

opinions_vecs = tfidf_vectorizer.fit_transform(document_df['opinions'])
print(opinions_vecs.toarray().shape) # (문서 수, 특성 수)
print(opinions_vecs)                 # 희소행렬 출력
````

````python
tfidf_vectorizer.vocabulary_ # 단어 사전
````

KMeans 군집화

````python
from sklearn.cluster import KMeans

kmeans = KMeans( 
    n_clusters = 4,    # 군집 개수
    max_iter = 5000,  # 반복 횟수
    random_state = 0  # 시드 고정
)

opinions_label = kmeans.fit_predict(opinions_vecs) 
document_df['cluster'] = opinions_label
document_df
````

````python
# 클러스터 중심 키워드(상위 TF-IDF) 추출
centers = kmeans.cluster_centers_  #각 클러스터별 중심 벡터
print(centers.shape)               # (클러스터 수, 특성 수)

centroid_arg_idx = centers.argsort()[:,::-1]  # 각 클러스터별 특성값 기준 내림차순하여 인덱스 반환
top_20 = centroid_arg_idx[ : : 20]            # 클러스터별 상위 20개 인덱스
feature_names = tfidf_vectorizer.get_feature_names_out()
feature_names[top_20]
````
