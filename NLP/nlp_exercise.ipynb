```python
from collections import Counter

# 샘플 코퍼스
corpus = ["자연어 처리는 재미있다.", "자연어 처리는 컴퓨터 과학의 한 분야다.", "데이터 분석은 자연어 처리와 밀접하다."]

# 단어 단위로 코퍼스를 분리
tokenized_corpus = [sentence.split() for sentence in corpus]
flat_tokens = [word for sentence in tokenized_corpus for word in sentence]

# 단어 빈도 분석
word_freq = Counter(flat_tokens)
print("단어 빈도 분석 결과:", word_freq)
```

```python
import re

# 원시 텍스트
raw_text = "Hello!!! 자연어 처리 is FUN.   Do you agree? :)"

# 불필요한 기호 제거
clean_text = re.sub(r"[^\w\s]", "", raw_text)

# 소문자 변환
clean_text = clean_text.lower()

# 공백 제거
clean_text = " ".join(clean_text.split())

print("처리된 텍스트:", clean_text)
```

```python
import re
import string
from nltk.tokenize import word_tokenize

raw_text = "자연어 처리는 데이터 과학의 한 분야다! 여러 전처리가 필요하다."

# 특수문자 제거
clean_text = raw_text.translate(
    str.maketrans("", "", string.punctuation)
)

# 숫자 제거: 역슬래시 하나만 사용
clean_text = re.sub(r"\d+", "", clean_text)

# 토큰화 데이터 설치가 필요 없는 방식
tokens = clean_text.split()

# 한국어 불용어 직접 정의
stop_words = {
    "은", "는", "이", "가",
    "을", "를", "의", "에",
    "한", "하다"
}

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]

print("전처리 결과:", filtered_tokens)
```

```python
from konlpy.tag import Okt

# 텍스트
text = "자연어 처리는 정말 재미있다!"

# 형태소 분석
okt = Okt()                 
morphs = okt.morphs(text)  # 최소 의미 단위 리스트 분리
pos = okt.pos(text)        # (형태소, 품사) 튜플 리스트로 품사 태깅 작업

print("형태소 분석 결과:", morphs)
print("품사 태깅 결과:", pos)
```

```python
from gensim.models import FastText   # 서브워드(n-그램) 기반 단어 임베딩 모델 FastText

# 1) 토큰화된 문장(리스트의 리스트) 준비
sentences = [
    ["자연어", "처리", "재미있다"],                          # 문장1 토큰 리스트
    ["파이썬", "으로", "자연어", "처리", "가능하다"],          # 문장2 토큰 리스트
    ["자연어", "처리", "공부", "하자"]                         # 문장3 토큰 리스트
]

# FastText: 단어를 '문자 n-그램(서브워드)' 조합으로도 표현해 OOV(미등장 단어) 벡터 생성에 강함
model = FastText(      # 2) FastText 학습(훈련)
    sentences,         # 학습 데이터(토큰화된 문장들)
    vector_size=50,    # 단어 벡터 차원 수
    window=3,          # 주변 문맥 범위(좌우 3개)
    min_count=1,       # 최소 등장 횟수(1 이상 단어만 사용)
    sg=1,              # 1=Skip-gram(중심→주변), 0=CBOW(주변→중심)
    min_n=2,           # 서브워드(문자 n-그램) 최소 길이
    max_n=5,           # 서브워드(문자 n-그램) 최대 길이
    epochs=50          # 전체 데이터 반복 학습 횟수
)

# 3) 단어 벡터 확인 : '자연어' 임베딩 벡터(50차원) 중 앞 10개 출력
print("자연어 벡터 일부:", model.wv["자연어"][:10])

# 4) 유사 단어(topn) : 코사인 유사도가 높은 단어 TOP3
print("자연어와 비슷한 단어:", model.wv.most_similar("자연어", topn=3))

# 5) 단어 간 유사도(코사인) : 두 단어 벡터의 코사인 유사도 스칼라 출력
print("similarity(자연어, 처리):", model.wv.similarity("자연어", "처리"))

# 6) OOV 예시: 학습에 없던 단어도 서브워드로 벡터 생성 가능
# "자연어처리" 미등장 단어라도 문자 n-그램으로 벡터 추정(앞 5개 출력)
print("OOV 단어 벡터 생성 가능?", model.wv["자연어처리"][:5])
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer

# 데이터 준비
corpus = ["자연어 처리는 재미있다", "Python은 강력하다"]
labels = [0, 1]

# 벡터화 및 분류
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)
model = MultinomialNB()
model.fit(X, labels)

# 예측
print(model.predict(vectorizer.transform(["Python이 재미있다"])))
```

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression

# 데이터 준비
corpus = ["자연어 처리는 재미있다", "Python은 강력하다"]
labels = [[1, 0], [0, 1]]  # 다중 레이블

# 모델 정의
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)
model = OneVsRestClassifier(LogisticRegression())
model.fit(X, labels)

# 예측
print(model.predict(vectorizer.transform(["Python이 재미있다"])))
```

```python
from transformers import BertTokenizer, BertModel

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

text = "Natural language processing is fascinating."
inputs = tokenizer(text, return_tensors="pt")
outputs = model(**inputs)
print(outputs.last_hidden_state)
```

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')

input_text = "Artificial Intelligence is"
inputs = tokenizer(input_text, return_tensors="pt")  # 입력 토큰 ID
outputs = model.generate(inputs['input_ids'],        
                         max_length=20, 
                         num_return_sequences=1)

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
