Encoder-Only 모델 : 네이버 영화리뷰로 학습한 감성분석 모델

```python
from transformers import pipeline

classifier = pipeline(
    "text-classification",
    model="daekeun-ml/koelectra-small-v3-nsmc"
)
```

```python
# 감성분석 함수
def analyze_sentiment(texts):
    results = classifier(texts)

    label_map = {"0": "부정", "1": "긍정"}

    for text, result in zip(texts, results):
        print(f"입력 문장 : {text}")
        print(f"감성 : {label_map[result["label"]]}")
        print(f"확률 : {round(result['score'], 4)}")
        print()
```

```python
texts = ["재미있고 감동적인 실화", "시간이 아깝네요.", "대작같은 졸작"]

analyze_sentiment(texts)
```

# Decoder-Only 모델 : 언어 생성

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct"
)
```

```python
# 입력 문장을 받아 자연스럽게 이어서 생성하는 함수
def generate_text(text):
    messages = [ # 모델에게 전달할 입력 메시지
        {
            "role" : "user", # 사용자 메시지
            "content" : f"다음 문장을 자연스럽게 이어서 작성해줘. {text}"
        }
    ]

    result = generator(
        messages,
        max_new_tokens = 100, # 최대 생성 토큰 수
        do_sample = True,     # 확률 기반으로 다양한 문장 생성
        temperature = 0.7     # 생성 다양성 조절 (0~1). 1에 가까울수록 창의적
    )

    return result[0]["generated_text"][-1]["content"] # 생성된 마지막 답변 내용 출력
```

```python
texts = [
    "인공지능이 발전하게 되면 인간을",
    "인간이 살아가는 이유는",
    "한국 사람이 똑똑한 이유는",
    "집가고 싶다"
]

for text in texts:
    print(f"입력 : {text}")
    print(f"생성 : {generate_text(text)}")
    print()
```
