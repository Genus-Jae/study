```python
import numpy as np
import matplotlib.pyplot as plt

# 시그모이드 함수 : 입력값을 0~1 사이 값으로 압축해 반환하는 함수

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-10, 10, 100)     # -10부터 10까지 균등한 간격으로 100개 
y = sigmoid(x)                    # 시그모이드 적용

import matplotlib.pyplot as plt
plt.plot(x, y)
plt.title("Sigmoid Function")
plt.show()
```

x가 너무 크거나 작으면 출력이 1 또는 0 근처로 평평해진다.                    
-> 기울기가 거의 0에 가까워지면서 깊은 신경망에서 학습이 느려진다.                 
그래서 출력층에서만 사용한다.                                       

```python
def step_function(x):
    return np.array(x > 0, dtype=int)

x = np.linspace(-10, 10, 100)
y = step_function(x)
plt.step(x, y)
plt.title("Step Function")
plt.show()
```

```python
def relu(x):
    return np.maximum(0, x)

x = np.linspace(-10, 10, 100)
y = relu(x)
plt.plot(x, y)
plt.title("ReLU Function")
plt.show()
```

양수 영역에서는 기울기가 항상 1                                                                    
-> 기울기가 1이란 말은 입력이 조금 바뀌었을 때 출력도 그대로 같은 비율로 바뀐다.                               

lr (learning_rate) 값을 수정해서 모델의 학습률 조정하는데
최적화 함수는 기존 파라미터 - (lr * 기울기) 값으로 업데이트 때문에 역전파 시에 기울기가 줄어들지 않고 잘 전달된다.        

음수가 들어오면 기울기가 0 -> Dead ReLU 문제 발생 (업데이트가 되지 않는 현상)
이를 위한 해결책으로 Leaky ReLU, Parametric ReLU, ELU 등 사용한다.

은닉층 활성화함수의 표준
출력층에서도 사용은 가능하나 거의 사용하지 않음.


```python
def leaky_relu(x, alpha=0.1):
    return np.where(x >= 0, x, alpha * x)

x = np.linspace(-10, 10, 100)
y = leaky_relu(x)
plt.plot(x, y)
plt.title("Leaky ReLU Function")
plt.show()
```

Leaky ReLU는 음수에서도 기울기가 어느정도 유지되어 가중치 업데이트를 진행할 수 있다.

```python
def tanh(x):
    return np.tanh(x)

x = np.linspace(-10, 10, 100)
y = tanh(x)
plt.plot(x, y)
plt.title("Tanh Function")
plt.show()
```

|구분|Sigmoid|Tanh|
|---|---|---|
|출력범위| 0~1 | -1 ~ 1|
|중심값|발생|완화되었으나 발생|
|사용 위치|츌룍충 (이진 분류)| 은닉층 (relu에 밀림)|
