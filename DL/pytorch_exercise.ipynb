argmax(dim = k) : k번 차원에서 가장 큰 인덱스를 선택하고, 그 차원 제거

````python
#topk() : 상위 k개의 값과 해당 인덱스를 튜플로 반환

values, indices = torch.topk(x, k = 1, dim = -1)

print(values, values.size())
print(indices, indices.size())
````

topk(k = 1)로 설정하면 argmax(keepdim = True)와 같은 결과

````python
# topk를 이용해 정렬하기
target_dim = -1
values, indices = torch.topk(
    x,
    k = x.size(target_dim),
    largest = True
)

print(values, values.size())
````

 x [ i, j, :]을 큰 값 기준으로 내림차순 정렬
- 차원 수는 그대로 유지. 값의 순서만 바뀜
- 각 줄에서는 큰 값 -> 작은 값 순서로 정렬
- indices는 정렬되기 전 위치값
- 결과적으로 topk(k = 차원크기)를 주면 해당 차원을 기준으로 정렬한 효과를 얻는다.

````python
k = 1
values, indices = torch.sort(x, dim = -1, descending = True) # 마지막 차원을 기준으로 내림차순 정렬 (값, 원래위치)
values, indices = values[:,:,:k], indices[:,:,:k]            # 정렬된 결과 중 가장 큰 값 1개 선택(3,3,1)

print(values.squeeze(-1), values.squeeze(-1).size())         # (3,3,1) -> (3,3)
print(indices.squeeze(-1), indices.squeeze(-1).size())       # (3,3,1) -> (3,3)

````

sort + 슬라이싱 = topk 사용과 동일한 효과를 얻는다.

````python
import numpy as np

def perceptron(x1, x2, weights, bias):
    x = np.array([x1, x2])               # 입력값을 하나의 vector로 묶음
    w = np.array(weights)                # 가중치 리스트를 벡터로 변환
    return int(np.dot(x, w) + bias > 0)
    # return 1 if (w1*x1 + w2*x2 + b) > 0 else 0

# AND Gate
print("AND:", perceptron(1, 1, [0.5, 0.5], -0.7))

# NAND Gate (AND 연산을 반대로)
print("NAND:", perceptron(1, 1, [-0.5, -0.5], 0.7))

# OR Gate
print("OR:", perceptron(1, 0, [0.5, 0.5], -0.2))
````

AND NAND OR 연산은 단층 퍼셉트론에서 해결 가능


````python
def weighted_sum(inputs, weights, bias):
    return np.dot(inputs, weights) + bias

inputs = [1, 0]
weights = [0.8, -0.5]
bias = 0.2
output = weighted_sum(inputs, weights, bias)
print("Weighted Sum:", output)
````

````python
from sklearn.neural_network import MLPClassifier

# XOR 문제 해결
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 1, 1, 0]

mlp = MLPClassifier(hidden_layer_sizes=(5,10),   # 은닉층 1개층, 뉴런 2개
                    activation='relu',         # relu 활성화함수(가장 보편적으로 사용하는 활성화함수)
                    max_iter=1000)             # epochs
mlp.fit(X, y)

print("Prediction:", mlp.predict(X))
````

activation : 은닉층에서 사용할 활성화 함수
- 입력에 비선형성을 추가해 복잡한 패턴 인식하는 함수

    - relu : 가장 일반적으로 사용, 학습 빠름
    - Leakt Relu : 음수 영역에 작은 기울기 허용 (죽은 ReLU 문제 완화)
    - ELU, GELU : Transformer 계열에서 사용

-tanh : 출력 범위가 -1 ~ 1로 나오는게 의미있을 때 사용
    - 입력/출력이 0을 중심으로 대칭
    - RNN계열 (LSTM, GRU)   

- logistic : 시그모이드 함수 (초기 신경망)



-출력층에서 사용할 활성화 함수와 구별을 잘 해야 한다.
이진 분류 : sigmoid (출력범위 0~1) -> 확률로 해석
다중 분류 : softmax (출력범위 0~1. 총합이 1) -> 각 클래스에 대한 확률 분포
회귀: linear(없음) -> 예측값 그대로 사용
