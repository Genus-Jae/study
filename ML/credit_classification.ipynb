4-1 수치형 변수 변환

4-2. 학습, 테스트 데이터 분리

```python
# 전체 데이터에서 입력값을 만들기
X = df_clean.drop('Risk', axis = 1)
X
```

```python
# 전체 데이터에서 정답값 가져오기
y = df_clean['Risk']
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size = 0.2,
    random_state = SEED,
    stratify = y             # target의 비율을 테스트/학습 데이터에서 비슷하게 유지
)

print(len(X_train), len(y_train), len(X_test), len(y_test))
```

4-3. 정규화

```python
# MinMaxScaler() : 모든 특성값을 0~1 사이로 정규화해주는 스케일러

scaler = MinMaxScaler()

feature_names = X.columns.tolist()      # 특성 이름 리스트

numeric_features = ['Age', 'Credit amount', 'Duration']

# 학습 데이터를 기준으로 학습하고, 변환
X_train[numeric_features] = scaler.fit_transform(X_train[numeric_features]) 

# 테스트 데이터는 학습 데이터 기준으로 (최소값/최대값) 변환
X_test[numeric_features] = scaler.transform(X_test[numeric_features])
```

5-1. LogisticRegression

```python
lr_model = LogisticRegression(random_state = SEED)

# 학습
lr_model.fit(X_train, y_train)

print("LogisticRegression 학습 완료")
```

```python
# 각종 평가지표들을 한 번에 출력해서 확인하고 반환받는 함수
def check_metrics( model, model_name, X_train, X_test, y_train, y_test):
    y_pred_train = model.predict(X_train)
    y_pred_test =  model.predict(X_test)
    

    # zero_division = 0 : 특정 클래스를 하나도 예측하지 못한 경우 경고 대신 0점 처리
    # pos_label = 1 : 위험 고객을 얼마나 잘 맞췄는지를 기준으로 계산
    acc_train = accuracy_score(y_train, y_pred_train)
    precision_train = precision_score(y_train, y_pred_train, zero_division = 0)
    recall_train = recall_score(y_train, y_pred_train, zero_division = 0)
    f1_train = f1_score(y_train, y_pred_train, zero_division = 0, pos_label = 1)

    acc_test = accuracy_score(y_test, y_pred_test)
    precision_test = precision_score(y_test, y_pred_test, zero_division = 0)
    recall_test = recall_score(y_test, y_pred_test, zero_division = 0)
    f1_test = f1_score(y_test, y_pred_test, zero_division = 0, pos_label = 1)


    print(f"{model_name}의 학습 정확도 : {acc_train:.4f}")
    print(f"{model_name}의 학습  정밀도 : {precision_train:.4f}")
    print(f"{model_name}의 학습  재현율 : {recall_train:.4f}")
    print(f"{model_name}의 학습  F1 스코어 : {f1_train:.4f}")


    print('-'*50)


    print(f"{model_name}의 테스트 정확도 : {acc_test:.4f}")
    print(f"{model_name}의 테스트 정밀도 : {precision_test:.4f}")
    print(f"{model_name}의 테스트 재현율 : {recall_test:.4f}")
    print(f"{model_name}의 테스트 F1 스코어 : {f1_test:.4f}")

    return {
        'model' : model_name,
        'accuracy' : acc_test,
        'precision' : precision_test,
        'recall' : recall_test,
        'f1_score' : f1_test
    }

# 혼동행렬을 계산하고 시각화하는 함수
def plot_confusion_matrix(model, model_name, X_test, y_test):
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)

    sns.heatmap(cm, 
                annot = True, 
                fmt = 'd', 
                cmap = 'Blues', 
                xticklabels = ["good(0)","bad(1)"], 
                yticklabels = ["good(0)","bad(1)"]
    )

    plt.xlabel("예측")
    plt.ylabel("실제값")
    plt.title(f"{model_name}의 혼동 행렬")
    plt.show()
```

```python
model_results = [] # 모델별 테스트 성능 결과 저장
```

```python
result = check_metrics(lr_model, "Logistic Regression", X_train, X_test, y_train, y_test)
model_results.append(result)
```

```python
plot_confusion_matrix(lr_model, "Logistic Regression", X_test, y_test)
```

6. 클래스 불균형 문제
- 데이터 비율이 7:3 이라서 SMOTE를 사용해서 비율을 맞춰주고 진행하면 성능개선 여지가 있다.

```python
# 데이터 증강 기법 SMOTE 
smote = SMOTE(random_state = SEED)

# 주의점 : 학습 데이터만 증강 적용해야 한다. (테스트 데이터에 적용하면 평가점수가 공정하게 되지 않음)
X_train_over, y_train_over = smote.fit_resample(X_train, y_train)

print(f"증강된 X 데이터 수 : {X_train_over.shape[0]}, 증강된 y 데이터 수 : {y_train_over.shape[0]}")

print(pd.Series(y_train_over).value_counts())
```

```python
xgb_over_model = XGBClassifier(random_state = SEED)
xgb_over_model.fit(X_train_over, y_train_over)

result = check_metrics(xgb_over_model, "증강된 데이터 XGBClassifier", X_train, X_test, y_train, y_test)

model_results.append(result)
```

고도화 실습 과제
추론모델을 함수형으로 모듈화시켜 streamlit에서 입력받은 값으로 출력 진행해보기

```python
import optuna
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score # 또는 f1_score 등

def objective(trial):
    # 튜닝할 파라미터 범위 설정 (궁딸깍 범위 지정)
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 1000),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0)
    }
    
    # 모델 생성
    model = XGBClassifier(**params, random_state=42)
    model.fit(X_train_over, y_train_over)
    
    # 평가 (점수가 높을수록 좋게 설정)
    preds = model.predict(X_test)
    return accuracy_score(y_test, preds)
```

```python
study = optuna.create_study(direction='maximize') # 점수 최대화
study.optimize(objective, n_trials=20) # 20번만 돌려봄
```

```python
best_params = study.best_params
final_model = XGBClassifier(**best_params, random_state=42)
final_model.fit(X_train_over, y_train_over)

# 결과 확인
result = check_metrics(final_model, "Optuna 튜닝 XGB", X_train, X_test, y_train, y_test)
model_results.append(result)
```
