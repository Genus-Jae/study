````python
g_variable = "global variable"

def modify_global():
    global g_variable
    g_variable = "global modified"

print(g_variable)    # global variable
modify_global()
print(g_variable)    # global modified
````

````python
def outer_function():
    variable = "outer variable"

    def inner_function():
        nonlocal variable
        variable = "modified"

    print(variable)
    inner_function()
    print(variable)

outer_function()
````

````python
class Character:
    def __init__(self, nickname, type):
        self.nickname = nickname #public attribute(속성)
        self.__type = type # private 속성

    def get_type(self):
        return self.__type

    def set_type(self, type):
        character_types = ['전사','법사','힐러']
        if type in character_types:
            self.__type = type
        else:
            raise ValueError('잘못된 캐릭터 타입입니다')

my_character = Character("카리바라","전사")

print(my_character.nickname)
#print(my_character.__type) --> error

print(my_character.get_type())

my_character.set_type("법사")
print(my_character.get_type())

my_character.set_type("마법사")
````

````python
class person:
    national = 'korea'

    def greeting(self):
        return 'Hello. This is Python'

class Learner:
    def greeting(self):
        return ' Hello. I am Learner'

    def learn(self):
        return ' I am Learning Python'

class Student(Person, Learner):
    pass

student = Student()
print(student.greeting())
print(student.learn())
````

````python
class Person:
    national = 'korea'
    
    def greeting(self):
        return 'Hello. This is Python'

class Learner(Person):
    def __init__(self, subject):
        self.subject = subject
    
    def learn(self):
        return 'I am learning ' + self.subject

class Student(Learner):
    def __init__(self, name, subject):
        super().__init__(subject)
        self.name = name
    
    def greeting(self):
        return 'Hello. My name is ' + self.name

student = Student('카피바라', 'python')
print(student.greeting())    # Hello. My name is 카피바라
print(student.learn())       # I am learning python
````

````python
class Animal:
    def do(self):
        print('👉👈')
class Dog(Animal):
    def do(self):
        print('🐶🐶🐶')
class Duck(Animal):
    def do(self):
        print('🦆🦆🦆')
class Fish(Animal):
    pass    # do 호출시 부모 클래스 Animal#do가 호출된다.

for each in (Dog(), Duck(), Fish()):
    each.do()
````

````python
class Alarm:
    def do(self):
        print('⏰⏰⏰')

for each in (Dog(), Duck(), Fish(), Alarm()):
    each.do()
````
