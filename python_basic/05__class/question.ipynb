문제 4: 네임스페이스 테스트 - 전역과 로컬
카피바라는 이직 준비 도구를 사용하며, 전역 변수와 로컬 변수를 관리합니다. namespace_test 함수를 작성해 네임스페이스를 테스트하세요.
전역 변수 tool을 "연필"로 설정하세요.
 함수 namespace_test 내부에 로컬 변수 local_tool을 "키보드"로 설정하고, global 키워드를 사용해 전역 변수를 "펜"으로 수정하세요.
함수 내부에서 로컬 변수와 전역 변수의 값을 출력하세요.

```python
tool = '연필'

def namespace_test():
    local_tool = '키보드'
    global tool
    tool = '펜'

    print(local_tool)
    print(tool)

namespace_test()
```

문제 5: 네임스페이스 테스트 - 클래스와 인스턴스
카피바라는 이직 준비 중 클래스를 통해 도구를 관리합니다. Toolbox 클래스를 만들어 네임스페이스를 테스트하세요.
클래스 속성 tools를 ["노트북", "책상"]으로 초기화하세요.
생성자를 통해 인스턴스 속성 tools를 설정하고, 객체마다 다른 도구를 가질 수 있도록 설정하세요.
add_tool 메서드를 만들어 클래스 속성에 새 도구를 추가하세요.

```python
class Toolbox:
    tools = ['노트북','책상']

    def __init__(self):
        self.tools = []

    def add_tool(self, tool):
        self.tools.append(tool)

toolbox1 = Toolbox()            
toolbox2 = Toolbox()     

toolbox1.add_tool('스피커')
toolbox1.add_tool('키보드')
toolbox2.add_tool('마우스')
toolbox2.add_tool('핸드폰')

print(toolbox1.tools)
print(toolbox2.tools)

print(Toolbox.tools)
```

문제 6: 네임스페이스 테스트 - nonlocal
카피바라는 이직 준비 중 추가적인 정보가 필요합니다. 중첩 함수와 nonlocal 키워드를 사용해 바깥 함수의 변수를 수정하세요.
함수 nested_function을 작성하세요.
바깥 함수의 변수 status를 "작성 중"으로 초기화하세요.
중첩 함수에서 status를 "완료"로 수정하세요.

```python
def nested_function():
    status = '작성 중'

    def inner_function():
        nonlocal status
        status = '완료'
    
    print(status)
    
    inner_function()

    print(status)

nested_function()        
```

문제 4: 상속의 특성 통합 테스트
카피바라는 강사님들의 지도를 받으며 다양한 분야에서 경험을 쌓고 있습니다. 상속과 메서드 오버라이딩, 다중 상속을 종합적으로 테스트하세요.
Mentor 클래스:
속성 field를 생성자로 받아 저장합니다.
메서드 guide는 "저는 [field] 분야의 멘토입니다."를 반환합니다.
Lecturer 클래스:
속성 experience를 생성자로 받아 저장합니다.
메서드 teach는 "저는 [experience]년의 강의 경력을 가지고 있습니다."를 반환합니다.
Candidate 클래스:
Mentor와 Lecturer 클래스를 다중 상속받습니다.
guide 메서드를 오버라이드하여 "저는 [field] 분야의 강사이며, 이직 준비를 도와드립니다."를 반환합니다.

```python
class Mentor:
    def __init__(self, field):
        self.field = field

    def guide(self):
        return '저는 ' +self.field + ' 분야의 멘토입니다.'

class Lecturer:
    def __init__ (self, experience) :
        self.experience = experience

    def teach(self):
        return "저는 " + self.experience + "년의 강의 경력을 가지고 있습니다."

class Candidate(Mentor, Lecturer):
    def __init__(self, field, experience):
        Mentor. __init__(self, field)
        Lecturer. __init__(self, field)

    def guide(self):
        return "저는 " + self.field + " 분야의 강사이며, 이직 준비를 도와드립니다"

mentor = Mentor("인공지능")
lect = Lecturer(10) 
cand = Candidate('AI', 30)

print(mentor.guide())
print(lect.teach())

print(cand.teach())
print(cand.guide())
```
