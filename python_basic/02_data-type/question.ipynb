```python
from collections import deque

dq = deque()

dq.append("a")
dq.append("b")
dq.append("c")

print(dq)    # deque(['a', 'b', 'c'])
```

```python
dq.appendleft('x')
print(dq)   # deque(['x', 'a', 'b', 'c'])
```

```python
value = dq.pop()
print(value)    # c

left_value = dq.popleft()
print(left_value)    # x

print(dq)    # deque(['a', 'b'])
```
